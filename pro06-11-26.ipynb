```python
l1 = list(map(int, input("ENTER THE FIRST LIST: ").split(",")))
l2 = list(map(int, input("ENTER THE SECOND LIST: ").split(",")))

if len(l1) == len(l2):
    print("TWO LISTS ARE SAME LENGTH")
else:
    print("TWO LISTS ARE DIFFERENT LENGTH")
    
if sum(l1) == sum(l2):
    print("TWO LISTS HAVE SAME SUM")
else:
    print("TWO LISTS ARE DIFFERENT SUM")
    
if set(l1) & set(l2):
    print("COMMON VALUE ",set(l1) & set(l2))
else:
    print("NO COMMON VALUE ")
  
```
